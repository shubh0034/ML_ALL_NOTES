# **Handling Mixed Variables**

```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv("/content/titanic (1).csv")
```

```python
df.head()
```

```python
df['number'].unique()
```

```python
fig=df['number'].value_counts().plot.bar()
fig.set_title('Passengers travelling with')
```

```python
# extract numerical part
df['number_numerical']=pd.to_numeric(df['number'],errors='coerce',downcast='integer')
```

```python
# extracting categorical data
df['number_categorical']=np.where(df['number_numerical'].isnull(),df['number'],np.nan)
```

```python
df.head()
```

```python
df['Cabin'].unique()
```

```python
df['Cabin_num']=df['Cabin'].str.extract(r'(\d+)')#capture numerical part
df['Cabin_cat']=df['Cabin'].str[0]#capture categorical part
```

```python
df.head()
```

```python
df['Cabin_cat'].value_counts().plot(kind='bar')
```

```python
date['date']=pd.to_datetime(date['date'])
```

```python
date.info()
```

**1. Extract year**

```python
date['date_year']=date['date'].dt.year
date.head()
```

**2. Extract Month**

```python
date['date_month_no']=date['date'].dt.month
date.head()
```

```python
date['date_month_name']=date['date'].dt.month_name()
date.head()
```

**3. Extract Days**

```python
date['date_day']=date['date'].dt.day
date.head()
```

```python
date['date_dow']=date['date'].dt.dayofweek
date.head()
```

```python
date['date_day_name']=date['date'].dt.day_name()
date.head()
```

```python
date['date_is_weekend']=np.where(date['date_day_name'].isin(['Sunday','Saturday']),1,0)
date.drop(columns=['product_id','city_id','orders']).head()
```

```python
date['date_week'] = date['date'].dt.isocalendar().week
date.head()
```

```python
date['quarter'] = date['date'].dt.quarter
date.head()
```

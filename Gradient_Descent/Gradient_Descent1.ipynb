# **Making of our own class know as gdRegression for learning gradient descent**

```python
class GDRegressor:
  def __init__(self,learning_rate,epochs):
    self.m=73.43
    self.b=-120
    self.lr=learning_rate
    self.epochs=epochs

  def fit(self,x,y):
    for i in range(self.epochs):
       loss_slope=-2*np.sum(y-self.m*x.ravel()-self.b)
       step_size=loss_slope*self.lr
       self.b=self.b-step_size
       print(loss_slope,self.b)
    print(self.b)
```

```python
gd=GDRegressor(0.001,100)
```

```python
gd.fit(x,y)
```

# **Dono me se ek bhi constant na ho**

```python
x,y=make_regression(n_samples=100,n_features=1,n_informative=1,n_targets=1,noise=20,random_state=13)
```

```python
plt.scatter(x,y)
```

```python
lr=LinearRegression()
```

```python
lr.fit(x,y)
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
from sklearn.model_selection import cross_val_score
```

```python
np.mean(cross_val_score(lr,x,y,scoring='r2',cv=10))
```

```python
class GDRegressor:
  def __init__(self,learning_rate,epochs):
    self.m=100
    self.b=-120
    self.lr=learning_rate
    self.epochs=epochs

  def fit(self,x,y):
    for i in range(self.epochs):
       loss_slope_b=-2*np.sum(y-self.m*x.ravel()-self.b)
       loss_slope_m=-2*np.sum((y-self.m*x.ravel()-self.b)*x.ravel())
       step_size_b=loss_slope_b*self.lr
       step_size_m=loss_slope_m*self.lr
       self.b=self.b-step_size_b
       self.m=self.m-step_size_m
    print(self.m,self.b)

  def predict(self,x):
    return self.m*x+self.b
```

```python
gd=GDRegressor(0.001,50)
```

```python
gd.fit(x,y)
```

```python
gd.predict(x)
```

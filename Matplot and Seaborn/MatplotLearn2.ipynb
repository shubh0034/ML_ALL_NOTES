# **3D Surface Plot**

```python
x=[np.linspace(-10,10,100)]
y=[np.linspace(-10,10,100)]
```

```python
xx, yy=np.meshgrid(x,y)
```

```python
z=xx**2+yy**2
z.shape
```

```python
fig=plt.figure(figsize=(12,8))

ax=plt.subplot(projection='3d')

p=ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
z1=np.sin(xx)+np.cos(yy)
fig=plt.figure(figsize=(12,8))

ax=plt.subplot(projection='3d')

p1=ax.plot_surface(xx,yy,z1,cmap='viridis')
fig.colorbar(p1)
```

# **Contour Plot**

```python
fig=plt.figure(figsize=(12,8))

ax=plt.subplot()

p=ax.contour(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
fig=plt.figure(figsize=(12,8))

ax=plt.subplot()

p=ax.contourf(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
fig=plt.figure(figsize=(12,8))

ax=plt.subplot()

p=ax.contourf(xx,yy,z1,cmap='viridis')
fig.colorbar(p)
```

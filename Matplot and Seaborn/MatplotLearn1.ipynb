```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# **2-D Line plot**



*   Bivariate Analysis
*   categorical -> numerical and numerical -> numerical
*  Use case - Time series data



```python
# plotting a simple example
price=[48000,54000,57000,49000,47000,45000]
year=[2015,2016,2017,2018,2019,2020]
plt.plot(year,price)
```

```python
batsman=pd.read_csv('/content/sharma-kohli.csv')
batsman
plt.plot(batsman['index'],batsman['V Kohli'])
```

```python
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dotted',linewidth=3,marker='o',markersize=10,label='Rohit')
plt.plot(batsman['index'],batsman['RG Sharma'],color='red',linestyle='dashed',marker='D',label='Rohit')
plt.title("Rohit Sharma vs Virat Kohli Career Comparison")
plt.xlabel('year')
plt.ylabel('total runs in a year')
plt.legend()
plt.grid()
plt.show()
```

```python
price=[48000,54000,57000,49000,47000,45000,450000]
year=[2015,2016,2017,2018,2019,2020,2021]
plt.plot(year,price)
plt.ylim(40000,70000)
plt.xlim(2015,2020)
```

# **Bar Chart**

* Both for bi and uni variate analysis
* Numerical vs Categorical
* Use case - Aggregrate analysis of groups

```python
Childern=[50,20,30,40]
colors=['red','yellow','blue','green']
plt.bar(colors,Childern,color='Red')
```

```python
plt.barh(colors,Childern,color='Red')
```

```python
df=pd.read_csv('/content/batsman_season_record.csv')
df
```

```python
plt.bar(df['batsman'],df['2015'])
```

```python
plt.bar(np.arange(df.shape[0])-0.2,df['2015'],width=0.2)
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2)
plt.bar(np.arange(df.shape[0])+0.2,df['2017'],width=0.2)
plt.xticks(np.arange(df.shape[0]),df['batsman'])
plt.show()
# if name is too big to adjust then just use rotate option in xtricks
```

```python
# stacked bar chart
plt.bar(df['batsman'],df['2015'],label='2015')
plt.bar(df['batsman'],df['2016'],bottom=df['2015'],label='2016')
plt.bar(df['batsman'],df['2017'],bottom=df['2016']+df['2015'],label='2017')
plt.legend()
plt.show()
```

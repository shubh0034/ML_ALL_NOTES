# **1. Relational Plot**

* to see the statistical relational between 2 or more variables.
* Bivariate analysis

Plots under this analysis

* Scatterplot
* Lineplot

```python
# scatter plot
tips=sns.load_dataset('tips')
tips
```

```python
#ye abhi axis level function h
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex',style='time',size='size')
```

```python
#ye figure level h
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',hue='sex',style='time',size='size')
```

```python
# lineplot
gap=px.data.gapminder()
gaps=gap[gap['country'] =='India']
gaps
```

```python
sns.lineplot(data=gaps,x='year',y='lifeExp')
```

```python
sns.relplot(data=gaps,x='year',y='lifeExp',kind='line')
```

```python
temp=gap[gap['country'].isin(['India','Brazil','Germany'])]
temp
```

```python
sns.relplot(data=temp,x='year',y='lifeExp',hue='country',kind='line',style='continent')
```

```python
# facet plot -> it will only work with fig level function
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',col='sex',row='smoker')
```

```python
# col wrap
sns.relplot(data=gap,x='lifeExp',y='gdpPercap',kind='scatter',col='year',col_wrap=4)
```

# **2. Distribution Plots **

* used for univariate analysis
* used to find out the observation
* Range of the observation
* Central Tendency
* is the data bimodal
* are there outliers

Plots under distribution plots
* histplot
* kdeplot
* rugplot

fig plot-> displot

axis plot -> above mention


```python
# histplot
sns.histplot(data=tips,x='total_bill')
```

```python
sns.displot(data=tips,x='total_bill',kind='hist',bins=20)
```

```python
sns.displot(data=tips,x='day',kind='hist',bins=20)
```

```python
sns.displot(data=tips,x='tip',kind='hist',bins=20,hue='sex',element='step')
```

```python
titanic=sns.load_dataset('titanic')
titanic
```

```python
sns.displot(data=titanic,x='age',kind='hist',element='step',hue='sex')
```

```python
sns.displot(data=titanic,x='age',kind='hist',element='step',col='sex')
```

```python
# kdeplot
# Rather than using dicrete bins, a KDE plot smooths the observation with a Gaussian Kernel, producing a continuous density estimate
sns.kdeplot(data=tips,x='total_bill')
```

```python
sns.displot(data=tips,x='total_bill',kind='kde',hue='sex',fill=True)
```

```python
# Rugplot
# Plot marginal distribution by drawing ticks along the x and y
# This function is intented to complement other plots by showing the location of indivisual observation in an unobtrusive way
sns.kdeplot(data=tips,x='total_bill')
sns.rugplot(data=tips,x='total_bill')
```

```python
# bivariate histogram
sns.histplot(data=tips,x='total_bill',y='tip')
```

```python
# bivariate kdeplot
sns.kdeplot(data=tips,x='total_bill',y='tip')
```

# **3. Matrix Plot**


*   HeatMap
*   Clustermap



```python
# Heatmap
temp=gap.pivot(index='country',columns='year',values='lifeExp')
plt.figure(figsize=(15,15))
sns.heatmap(temp)
```

```python
temp=gap[gap['continent']=='Europe'].pivot(index='country',columns='year',values='lifeExp')
plt.figure(figsize=(15,15))
sns.heatmap(temp,annot=True,cmap='viridis')
```

```python
# ClusterMap
# Plot a matrix dataset as a hierarchically-clustered heatmap
iris=px.data.iris()
```

```python
sns.clustermap(iris.iloc[:,[0,1,2,3]])
```

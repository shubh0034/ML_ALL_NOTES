```python
# looping in groups
df=pd.DataFrame(columns=movies.columns)
df_list=[]
for group,data in genres:
  # print(type(group),type(data))
  max_row=data[data['IMDB_Rating']==data['IMDB_Rating'].max()]
  df_list.append(max_row)
df=pd.concat(df_list)
df
```

```python
# split(apply) combine
genres.apply(min)
```

```python
# find no. of movies starting with a for each group
def foo(group):
  return group['Series_Title'].str.startswith('A').sum()
```

```python
genres.apply(foo,include_groups=False)
```

```python
# find ranking of each movie in the group according to IMDB score
def rank_value(group):
  group['genre_rank']=group['IMDB_Rating'].rank(ascending=False)
  return group
```

```python
genres.apply(rank_value,include_groups=False)
```

```python
# find normalized IMDB rating group wise
def normal(group):
  group['norm_rating']=(group['IMDB_Rating']-group['IMDB_Rating'].min())/(group['IMDB_Rating'].max()-group['IMDB_Rating'].min())
  return group
```

```python
genres.apply(normal)
```

```python
ipl=pd.read_csv('deliveries.csv')
ipl.head()
```

```python
ipl.shape
```

```python
# find the top ten batsman in terms of runs score
batsman=ipl.groupby('batsman')
batsman['batsman_runs'].sum().sort_values(ascending=False).head(10)
```

```python
# find the name of the batsman with most no. of sixes
six=ipl[ipl['batsman_runs']==6]
six.groupby('batsman')['batsman'].count().sort_values(ascending=False).head(5)
```

```python
# find V Kohli's record against all the team
team_df=ipl[ipl['batsman']== 'V Kohli']
team_df.groupby('bowling_team')['batsman_runs'].sum().reset_index()
```

```python
# Create a function that can return the highest score of any batsman
team_df=ipl[ipl['batsman']== 'V Kohli']
team_df.groupby('match_id')['batsman_runs'].sum().sort_values(ascending=False).head(1).values[0]
```

```python
def highest(batsman):
  team_df=ipl[ipl['batsman']== batsman]
  return team_df.groupby('match_id')['batsman_runs'].sum().sort_values(ascending=False).head(1).values[0]
```

```python
highest('MS Dhoni')
```

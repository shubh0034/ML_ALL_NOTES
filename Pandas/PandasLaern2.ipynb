```python
# using list
student_data=[
    [100,80,10],
    [90,70,7],
    [120,100,14],
    [80,50,2]
]

pd.DataFrame(student_data,columns=['iq',"marks","package(in lakhs)"])
```

# **Filtering DataFrame**

```python
ipl=pd.read_csv('/matches.csv')
ipl
```

```python
# city=ipl['city']=='Delhi'
# delhi=ipl[city]
# delhi[['date','winner']]
ipl[ipl['city']=='Delhi'][['date','winner']]
```

```python
ipl[ipl['result']!='normal'][['date','team1','team2','result']].shape[0]
```

```python
ipl[(ipl['city']=='Kolkata') & (ipl['winner']=='Chennai Super Kings')].shape[0]
```

```python
tosswinner=ipl[ipl['toss_winner'] == ipl['winner']].shape[0]
totalmatches=ipl.shape[0]
(tosswinner*100)/totalmatches
```

```python
# for more than one option in the particular column
# or movies['Genre'].str.contains('Action)
movies[(movies['Genre'].str.split('|').apply(lambda x: 'Action' in x)) &(movies['Budget']> 30)]
```
